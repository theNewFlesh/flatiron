```python
from flatiron.core.dataset import Dataset
import cv_depot.api as cvd
import pandas as pd
import re
import os
import shutil
```

```python
def chunk_to_images(data, pattern):
    f = 0
    for chunk in data:
        for i in range(chunk.shape[0]):
            img = cvd.Image.from_array(chunk[i])
            filepath = pattern.format(frame=f)
            img.write(filepath)
            f += 1
```

```python
src = '/mnt/storage/projects/unet001/dset001/p-unet001_s-dset001_d-glom_v001'
data = Dataset.read_directory(src)
```

```python
asset = 'p-unet001_s-dset002_d-glom_v001'
tgt = f'/mnt/storage/projects/unet001/dset002/{asset}'
shutil.rmtree(tgt)
tgt = f'{tgt}/data'
os.makedirs(tgt, exist_ok=True)
pattern = f'{tgt}/{asset}_f{{frame:04d}}.exr'
chunk_to_images(data, pattern)
```

```python
data = pd.DataFrame()
data['filepath_relative'] = os.listdir(tgt)
data['filepath_relative'] = data.filepath_relative.apply(lambda x: f'data/{x}')
data['asset_path'] = f'/mnt/storage/projects/unet001/dset002/{asset}'
src1 = f'/mnt/storage/projects/unet001/dset002/{asset}/{asset}.csv'
data.to_csv(src1, index=None)
```
